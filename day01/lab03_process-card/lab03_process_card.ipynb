# 실습 3: 내 공정 카드 만들기

### 용어 풀이 - 반도체 8대 공정

| 순서 | 공정 | 한 줄로 |
|---|---|---|
| 1 | 웨이퍼 제조 | 실리콘으로 얇은 원판(웨이퍼)을 만들어 표면을 거울처럼 다듬는다 |
| 2 | 산화 | 웨이퍼 표면에 얇은 산화막을 입혀 보호막이자 절연막으로 삼는다 |
| 3 | 포토 (노광) | 빛으로 회로 무늬를 웨이퍼에 찍는다. 도장 찍기에 가깝다 |
| 4 | 식각 (에칭) | 찍은 무늬를 따라 필요 없는 부분을 깎아낸다 |
| 5 | 증착·이온주입 | 얇은 막을 입히고(증착), 전기 성질을 바꿀 입자를 박아 넣는다(이온주입) |
| 6 | 금속배선 | 만들어진 소자들을 금속 선으로 연결한다 |
| 7 | EDS (Electrical Die Sorting) | 자르기 전에 칩 하나하나에 전기를 흘려 합격·불합격을 판정한다 |
| 8 | 패키징 | 합격한 칩을 잘라내 포장하고 외부와 연결되게 만든다 |

2~6번은 한 번으로 끝나지 않고 층을 쌓는 만큼 여러 번 반복된다.

EDS는 재료 분야에서 쓰는 EDS(에너지 분산 분광법, Energy Dispersive Spectroscopy)와 철자만 같고 전혀 다른 말이다. 여기서는 전기를 흘려 칩을 골라내는 검사를 뜻한다.

### 목록에는 없지만 계속 나오는 것

| 말 | 뜻 |
|---|---|
| 평탄화 (CMP) | 층을 쌓다 보면 울퉁불퉁해진다. 갈아서 평평하게 만든다 |
| 세정 | 공정 사이사이 남은 찌꺼기와 입자를 씻어낸다 |

이 둘은 8대 공정 목록에 이름이 없지만, 실제로는 공정 사이사이마다 반복해서 들어간다.

### 자주 나오는 말

| 말 | 뜻 |
|---|---|
| 웨이퍼 | 회로를 쌓아 올리는 얇은 원판 |
| 챔버 | 공정이 실제로 일어나는 밀폐된 방 |
| 수율 | 만든 것 중 쓸 수 있는 것의 비율 |
| 계측 | 값을 재는 것. 잰 값이 곧 데이터의 한 칸이 된다 |
| 전공정 / 후공정 | 웨이퍼 위에서 회로를 만드는 앞쪽(1~6) / 잘라내 포장하는 뒤쪽(8) |

## Step 2. AI에게 카드 재료 받기

### 증착 (Deposition) 공정 카드

**1. 하는 일** — 웨이퍼 표면에 원하는 물질을 얇은 막 형태로 입힌다.

**2~3. 측정값과 정상 범위를 벗어났을 때 생기는 문제**

| 항목 | 단위 | 벗어나면 생기는 문제 |
|---|---|---|
| 증착 두께 (Thickness) | Å 또는 nm | 얇으면 절연·보호 기능 부족, 두꺼우면 단차 커버리지·후속 공정 문제 |
| 증착률 (Deposition Rate) | Å/min | 낮으면 생산성 저하, 높으면 막질 치밀도 저하·두께 재현성 저하 |
| 챔버 온도 | °C | 낮으면 막의 접착력·결정성 저하, 높으면 막 내부 응력 증가·웨이퍼 휨 |
| 챔버 압력 | mTorr | 낮으면 증착 속도 저하, 높으면 균일도 저하·이물(파티클) 증가 |
| 가스 유량 | sccm | 부족·과다 시 막의 조성비가 틀어져 전기적 특성 불량 |

**4. 확실하지 않음 / 장비·현장마다 다름**

- CVD(화학기상증착)인지 PVD(물리기상증착, 스퍼터링 등)인지에 따라 측정 항목·단위가 달라짐 — 위 표는 공통 항목 기준
- 입히는 물질(산화막, 금속, 폴리실리콘 등)에 따라 정상 범위가 크게 다름
- 정확한 수치는 장비 레시피·스펙시트 확인 필요


## Step 3. 받은 답을 내 눈으로 검토하기

### 증착 두께 — 더 알아보기

| 질문 | 내용 |
|---|---|
| 실제 단위 | 보통 Å(옹스트롬) 또는 nm으로 잰다. 막이 두꺼운 편이면 μm을 쓰기도 한다 *(막 종류·두께대에 따라 관행이 달라 확실하지 않음)* |
| 왜 중요하게 보나 | 두께가 절연/전도 같은 소자 성능을 직접 좌우하고, 다음 공정(식각·CMP)의 여유(마진)에도 영향을 주기 때문에 배치마다 계측해서 관리한다 |
| 장비·회사마다 다른가 | 계측 방식(엘립소미터, 반사분광 등)과 목표 두께·관리 기준이 회사·제품마다 다르다 *(구체적 수치는 사내 스펙 문서를 봐야 해서 확실하지 않음)* |


### 증착 두께 — 심화 정리

**측정 원리와 단위**

| 구분 | 내용 |
|---|---|
| 단위 | 옹스트롬(Å, 0.1 nm) 또는 나노미터(nm)로 표기. 수 Å~수백 nm대 박막이 일반적이며, 배선층처럼 두꺼운 막은 μm 단위를 쓰기도 한다 |
| 대표 계측법 | 엘립소미터(Ellipsometry, 편광 변화를 이용) 또는 반사분광법(Reflectometry, 박막 간섭을 이용) — 비접촉·비파괴 방식이라 인라인(in-line) 계측에 주로 쓰인다 |
| 보조 계측법 | 단면 SEM(전자현미경) 관찰, 전도성 막은 4-point probe로 면저항을 재서 두께로 환산 — 정밀 검증이나 초기 레시피 세팅 때 사용 *(선택 기준은 현장마다 달라 확실하지 않음)* |

**현장에서 중요하게 보는 이유**

- 두께는 절연막의 내압·정전용량, 배선의 저항 같은 소자의 전기적 특성을 직접 결정하는 변수다
- 목표치에서 벗어나면 뒤이은 식각·CMP 공정의 기준(target)과 어긋나, 오차가 다음 공정으로 누적된다
- 웨이퍼 내 위치별 두께 편차(균일도)는 다이(die)마다 성능 산포로 이어져 수율에 영향을 준다
- 이런 이유로 대부분 배치마다 인라인 계측을 하고, 통계적 공정관리(SPC) 관리한계를 벗어나면 경보를 띄운다

**장비·회사마다 달라지는 부분**

- 계측 장비의 제조사·모델에 따라 측정 원리(엘립소미터 vs 반사분광)와 정밀도가 다르다
- 목표 두께와 관리 상하한(spec)은 제품·레이어별로 회사 내부 기준으로 정해지며, 공개된 공통 표준값은 없다
- CVD·PVD·ALD 등 증착 방식에 따라 두께 재현성 수준과 계측 주기가 달라질 수 있다

*(이 셀은 일반적으로 알려진 경향을 정리한 것으로, 구체적 수치·사내 기준은 확인이 필요해 확실하지 않은 내용을 포함한다)*

## Step 5. 590이라는 숫자 앞에서

내 공정 하나에서만 측정 항목이 \[5\]개.<br>
공정이 8개니까 -> [5] x 8 = [40]개.<br>
층을 쌓느라 같은 공정을 여러 번 반복한다. 여기서는 20번이라고 치고 계산해본다.<br>
-> [40] x 20 = [800]개. 590이 나오는 게 이상하지 않다.

확인 결과 : [조건에 따라 다르다] 
<br>- 막의 재질, 막 두께, 다층 구조 여부, n·k(광학상수) 측정 필요 여부 등을 기준으로 선택함
<br>- 판단 기준 순서: 막 종류 > 두께 > 다층 구조 여부 > n,k 필요 여부 > ...

| 계측 방식                   | 주로 측정하는 값        | 선택하는 기준                          | 주 사용 목적     | 비고                    |
| ----------------------- | ---------------- | -------------------------------- | ----------- | --------------------- |
| **분광 엘립소미터(SE)**        | 두께(nm), **n, k** | 초박막, 다층막, 광학상수(n, k)가 필요한 경우 | 연구개발 + 양산   | 가장 많이 사용하는 비접촉 광학 계측  |
| **반사분광(Reflectometry)** | 두께(nm, μm)       | 단층 또는 단순 투명막, 빠른 측정이 필요한 경우      | 양산 인라인      | 속도가 빠르지만 복잡한 다층막에는 한계 |
| **프로파일러(단차측정)**         | 단차 높이(nm, μm)    | 막 일부를 제거하여 단차가 있는 경우             | 공정 검증       | 광학식이 아닌 실제 높이 측정      |
